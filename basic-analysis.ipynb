## Investigate inappropriate values

### Investigate cost by country and currency

```python
# Summarize cost by country and currency
cost_summary = (
    df.groupby(["Country Code", "Currency"])["Average Cost for two"].agg(
        restaurant_count="count",
        median_cost="median",
        mean_cost="mean",
        max_cost="max",
    ).sort_values(by="max_cost", ascending=False).reset_index()
)

print(cost_summary.to_string(index=False))
```

```python
f"{8652/9551*100:.2f}%"
```

```python
cost_outliers = df.nlargest(
    10, "Average Cost for two"
)[["Restaurant Name",
   "Country Code",
   "City",
   "Average Cost for two",
   "Currency",
   "Price range",
]]

print(cost_outliers.to_string(index=False))
```

**Conclusion:**
- From the analysis above, we observe that **the majority of the restaurants (90.59%) are based in India**, and the costs varied extremely between countries so the cost it should be analyzed in accordance with its locality.
- The country with the highest cost/price is Indonesia, this is due to Indonesia Rupiah being significantly weaker compared to a global currency such as the Dollar. Nevertheless, the number **800,000 is not an outlier but rather a cost presented in a weak currency**, there is no concrete reason to remove it.

### Investigate zero coordinates

```python
zero_coords = df[(df["Latitude"] == 0) | (df["Longitude"] == 0)]
print(f"Records with zero coordinates: {len(zero_coords)}")

print("\nBy country and city:")
print(zero_coords.groupby(["Country Code", "City"]).size().sort_values(ascending=False).to_string())

print("\nSample records:")
print(zero_coords[[
    "Restaurant Name",
    "Country Code",
    "City",
    "Address",
    "Longitude",
    "Latitude",
    ]].head(20).to_string(index=False)
)
```

**Conclusion**
- Since most of the restaurants with zero coordinates in either longitude or latitude are mostly in the country with Country Code 1, from which cities we can infer that it's India, it's highly likely that the zero coordinates are actually placeholders instead of actual coordinates. And the coordinates of (0,0) belongs to the Null Island off the coast of Gulf of Guinea.

### Investigate zero ratings

```python
zero_rating = df["Aggregate rating"] == 0
not_rated = df["Rating text"] == "Not rated"

print("Not rated with zero rating:", (zero_rating & not_rated).sum())
print("\nRating text/number combinations:")
print(pd.crosstab(zero_rating, not_rated))
print("\nRating text/color combinations:")
print(pd.crosstab(df["Rating text"], df["Rating color"]))
```

**Conclusion**
- All of zero ratings are correctly classified as "Not Rated", ensuring the validity of the ratings.
- All colors are in accordance with its rating text value

### Investigate zero costs

```python
print(
    df.loc[df["Average Cost for two"] == 0, [
        "Restaurant Name", 
        "Country Code",
        "City",
        "Average Cost for two",
        "Currency",
        "Price range",
    ]].to_string(index=False)
)
```

**Conclusion**
- The restaurants with zero costs are all have the price range of 1, this could either be missing or unrecorded values. Nevertheless, the value is not invalid so we will preserve the values but **exclude them from any cost analysis**.
